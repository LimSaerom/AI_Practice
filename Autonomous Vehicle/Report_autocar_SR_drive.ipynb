```python
#1 패키지 불러오기_25.4.1
from pop import Pilot
autocar = Pilot.AutoCar()
import time
```

```python
# 이동 코드_U턴
autocar.forward(30)         #속도 30
time.sleep(10)              #10초간 전진

autocar.backward(30)
autocar.setLamp(1,1)        # (전방, 후방)
time.sleep(5)               #5초간 후진
autocar.setLamp(0,0)        #멈추기전 라이트 꺼짐

autocar.steering=-1         #좌측으로 방향 전환
autocar.forward(10)         #속도 줄여서 유턴

start_time = time.time()     #좌회전 시작 시간 기록

while True:
    elapsed_time = time.time() - start_time  # 경과 시간 계산
    if elapsed_time >= 5 :
        autocar.steering=0  # 중앙
        break

    time.sleep(0.1)

autocar.backward(10)         #속도 10
time.sleep(1)
autocar.stop()
```
